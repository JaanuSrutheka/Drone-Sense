```python
import librosa
import librosa.display
import matplotlib.pyplot as plt
import numpy as np

drone_path = "../data/samples/sample_drone.wav"
noise_path = "../data/samples/sample_background_noise.wav"

drone, sr_drone = librosa.load(drone_path, sr=None)
noise, sr_noise = librosa.load(noise_path, sr=None)

print("Drone duration:", len(drone) / sr_drone)
print("Drone sample rate:", sr_drone)

print("Background duration:", len(noise) / sr_noise)
print("Background sample rate:", sr_noise)

plt.figure(figsize=(12, 4))
librosa.display.waveshow(drone, sr=sr_drone)
plt.title("Drone Audio")
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")
plt.show()
```
